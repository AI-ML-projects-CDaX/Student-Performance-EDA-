```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create the Dataset
data = {
    'Student': ['A', 'B', 'C', 'D', 'E'],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female'],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math'],
    'Marks': [78, 85, 92, 67, 88],
    'Attendance': [85, 90, 95, 70, 92]
}

df = pd.DataFrame(data)

print("\nStudent Data:\n")
print(df)

# Step 2: Average & Highest Marks
average_marks = df['Marks'].mean()
highest_marks = df['Marks'].max()

print("\nAverage Marks:", average_marks)
print("Highest Marks:", highest_marks)

# Step 3: Subject-wise Performance
subject_performance = df.groupby('Subject')['Marks'].mean()

print("\nSubject-wise Average Marks:\n")
print(subject_performance)

# Bar Chart for Subject-wise Performance
plt.figure()
subject_performance.plot(kind='bar')
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.show()

# Step 4: Correlation Analysis
correlation = df['Attendance'].corr(df['Marks'])
print("\nCorrelation between Attendance and Marks:", correlation)

# Scatter Plot
plt.figure()
plt.scatter(df['Attendance'], df['Marks'])
plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.show()

# Step 5: Gender-wise Performance
gender_performance = df.groupby('Gender')['Marks'].mean()

print("\nGender-wise Average Marks:\n")
print(gender_performance)

# Bar Chart for Gender-wise Performance
plt.figure()
gender_performance.plot(kind='bar')
plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")
plt.show()

print("\nAnalysis Completed Successfully!")
```
